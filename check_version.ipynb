# Environment Version Check


```python
import torch
torch_version = torch.__version__
cuda_is_available = torch.cuda.is_available()

print(f"PyTorch version: {torch_version}")
print(f"CUDA available: {cuda_is_available}")
if cuda_is_available:
    cuda_version = torch.version.cuda
    cudnn_version = torch.backends.cudnn.version()
    print(f"CUDA version: {cuda_version}")
    print(f"cuDNN version: {cudnn_version}")
```

```python
torch_build_config = torch.__config__.show()
print(torch_build_config)
```

```python
import subprocess
python_version_output = subprocess.check_output(
    ["python", "--version"],
    text=True,
    stderr=subprocess.STDOUT,
).strip()
print(python_version_output)
```

```python
import subprocess
gpp_version_output = subprocess.check_output(
    ["g++", "-v"],
    text=True,
    stderr=subprocess.STDOUT,
)
print(gpp_version_output)
```

```python
import importlib
def print_module_version(module_name: str, display_name: str) -> None:
    try:
        module = importlib.import_module(module_name)
        version = getattr(module, "__version__", "unknown")
        print(f"{display_name} version: {version}")
    except Exception as exc:
        print(f"{display_name} version: unavailable ({exc})")


MODULES_TO_CHECK = [
    ("flash_attn", "FlashAttention"),
    ("causal_conv1d", "CausalConv1d"),
    ("mamba_ssm", "Mamba-SSM"),
]

for module_name, display_name in MODULES_TO_CHECK:
    print_module_version(module_name, display_name)
```
